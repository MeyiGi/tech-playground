## Soru 3: Kargo kararı

**25 puan · 7 dakika · Hafta 3**

Kullanıcıdan alışveriş tutarını ve üyelik bilgisini alın.
Üyelik girdisi küçük harflerle tam olarak `evet` veya `hayir` olmalıdır.

Aşağıdaki örnek kurallara göre kargo ücretini belirleyin:

- Tutar negatifse veya üyelik bilgisi geçersizse yalnızca `Geçersiz giriş` yazın.
- Tutar **500 veya üzerindeyse** kargo ücretsizdir.
- Üye olanlarda tutar **200 veya üzerindeyse** de kargo ücretsizdir.
- Diğer geçerli durumlarda kargo ücreti **40** olur.

Kararınızda en az bir bileşik koşul kullanın. Geçerli girişte sonucu
`Kargo: 40.00` veya `Kargo: 0.00` biçiminde yazdırın.
Tutar için sayıya dönüştürülebilen bir metin girildiğini varsayın.

| Tutar | Üyelik | Beklenen sonuç |
| --- | --- | --- |
| 500 | hayir | Kargo: 0.00 |
| 499 | hayir | Kargo: 40.00 |
| 200 | evet | Kargo: 0.00 |
| 199 | evet | Kargo: 40.00 |
| -1 | evet | Geçersiz giriş |
| 300 | belki | Geçersiz giriş |

**Puanlama:** Girdiler 4, geçersiz giriş denetimi 5, kargo koşulları 10, çıktı 6 puan.


````python
# Soru 3 çözümünüzü buraya yazın.
amount = float(input("Amount: "))
member = input("Membership (evet/hayir): ")

if amount < 0 or member not in ["evet", "hayir"]:
    print("Geçersiz giriş")
elif amount >= 500 or (member == "evet" and amount >= 200):
    print("Kargo: 0.00")
else:
    print("Kargo: 40.00")
````

## Soru 4: Geçen notların özeti

**40 puan · 13 dakika · Haftalar 3–5**

`gecen_ozeti(notlar, esik=60)` adlı bir fonksiyon yazın.

- `notlar` bir sayı listesi olsun. Geçerli notların `0–100` aralığında olduğunu varsayın.
- Bir not **eşiğe eşit veya büyükse** geçmiş sayılsın. Varsayılan eşik `60` olsun.
- Bir **döngü** kullanarak yalnızca geçen notların toplamını ve adetini hesaplayın.
- Sonucu `(gecen_adet, gecen_ortalama)` biçiminde bir **demet** olarak döndürün.
- Liste boşsa veya geçen not yoksa `(0, None)` döndürün.
- Fonksiyon listeyi değiştirmesin, kullanıcıdan girdi istemesin ve ekrana yazmasın.
  Kısa bir docstring ile ne döndürdüğünü açıklayın.

| Çağrı | Beklenen dönüş değeri |
| --- | --- |
| `gecen_ozeti([45, 60, 75, 90])` | `(3, 75.0)` |
| `gecen_ozeti([45, 60, 75, 90], esik=80)` | `(1, 90.0)` |
| `gecen_ozeti([60])` | `(1, 60.0)` |
| `gecen_ozeti([20, 40])` | `(0, None)` |
| `gecen_ozeti([])` | `(0, None)` |

Fonksiyonu tanımladıktan sonra tablodan **en az iki örneği** fonksiyonun dışında
çağırıp sonuçlarını yazdırın. Bunlardan biri boş liste veya geçen not olmayan durum olsun.

**Puanlama:** Tanım ve parametreler 6, eşik koşulu 8, döngü ve biriktirme 10,
boş/geçen not olmayan durum 8, dönüş sözleşmesi ve yan etkisiz çalışma 8 puan.


````python
# Soru 4: Fonksiyon tanımınızı ve örnek çağrılarınızı buraya yazın.
def gecen_ozeti(notlar, esik=60):
    """Returns the count and average of passing grades."""
    count = 0
    total = 0

    for grade in notlar:
        if grade >= esik:
            count += 1
            total += grade

    if count == 0:
        return (0, None)

    average = total / count
    return (count, average)


print(gecen_ozeti([45, 60, 75, 90]))
print(gecen_ozeti([20, 40]))
print(gecen_ozeti([]))
````
